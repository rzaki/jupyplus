## Introduction

Whether we are summing values, accumulating data, or comparing elements, **loops are the heartbeat of software engineering**.

## C++11: Range-based for Loop

```c++
#include <vector>
#include <iostream>

std::vector<double> m_cashPayments = {100.0, 200.0, 150.0, 300.0};

for (const auto& payment : m_cashPayments) {
    std::cout << "Payment: " << payment << "\n";
}
```

## The Problem with Parallel Containers (C++11)

```c++
std::vector<int> m_timePeriods = {1, 2, 3, 4};

double presentValue(double payment, int period) {
    return payment / (1.0 + 0.05 * period);
}

// C++11 - Manual index (error-prone)
double old_way() {
    double total = 0.0;
    int i = 0;
    for (const auto& period : m_timePeriods) {
        total += presentValue(m_cashPayments[i], period);
        ++i;
    }
    return total;
}

std::cout << "Present Value (old way): " << old_way() << "\n";
```

## C++20: Better Index Scoping

```c++
// C++20 - Index scoped inside loop
double cpp20_way() {
    double total = 0.0;
    for (int i = 0; const auto& period : m_timePeriods) {
        total += presentValue(m_cashPayments[i], period);
        ++i;
    }
    return total;
}

std::cout << "Present Value (C++20): " << cpp20_way() << "\n";
```

## ✨ C++23: std::views::zip — The Modern Solution

```c++
#include <ranges>

// C++23 - Clean and elegant
double cpp23_way() {
    double total = 0.0;
    for (auto const& [payment, period] : std::views::zip(m_cashPayments, m_timePeriods)) {
        total += presentValue(payment, period);
    }
    return total;
}

std::cout << "Present Value (C++23 with zip): " << cpp23_way() << "\n";
```

## Takeaway

We have evolved from manipulating indices and pointers to **declaring what we want to iterate over**.

Modern C++ lets us write code that clearly expresses intent — with zero runtime overhead.
